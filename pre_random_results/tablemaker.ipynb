```python
import pandas as pd
import os

INPUT="summary.csv"

OUTDIR="tables"

os.makedirs(OUTDIR,exist_ok=True)

df=pd.read_csv(INPUT)

OUTLIERS=[None,"ZScore","IsolationForest","IQR"]

MODELS=df["model"].unique()

def format_score(row):

    return f"{row['mean']:.4f} ± {row['std']:.4f}"

for model in MODELS:

    mdf=df[df["model"]==model]

    datasets=sorted(mdf["dataset"].unique())

    rows=[]

    for d in datasets:

        row={"Dataset":d}

        sub=mdf[mdf["dataset"]==d]

        for o in ["None","ZScore","IsolationForest","IQR"]:

            if o=="None":

                filt=sub[sub["outlier"].isna()]

            else:

                filt=sub[sub["outlier"]==o]

            if len(filt)==0:

                row[f"{o}_score"]="NA"

                row[f"{o}_removed"]="NA"

                continue

            r=filt.iloc[0]

            row[f"{o}_score"]=format_score(r)

            if o=="None":

                row[f"{o}_removed"]=0

            else:

                row[f"{o}_removed"]=round(r["removed_pct"]*100,2)

        rows.append(row)

    table=pd.DataFrame(rows)

    table=table.sort_values("Dataset")

    table.to_csv(

        f"{OUTDIR}/{model}_table.csv",

        index=False
    )

print("Tables created in /tables")
```
