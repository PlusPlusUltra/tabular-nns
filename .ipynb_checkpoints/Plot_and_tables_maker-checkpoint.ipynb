```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

RESULTS_FILE = "results/summary.csv"

CLASSIFICATION = [
"cmc",
"connect-4",
"credit-g",
"kc1",
"splice",
"vehicle",
"phoneme",
"eucalyptus",
"eye_movements",
"house_16H",
"Bank_Customer_Churn_Dataset"
]

REGRESSION = [
"elevators",
"socmob",
"pol"
]

MODELS = ["mlp","xgboost","randomforest"]


########################################
# LOAD
########################################

df = pd.read_csv(RESULTS_FILE)

df["method"] = df["detector"].astype(str) + "_" + df["treatment"]


########################################
# IMPROVEMENT
########################################

def compute_improvement(group):

    baseline = group[
        (group.detector.isna()) |
        (group.detector=="None")
    ]

    if len(baseline)==0:
        return group

    base = baseline.iloc[0]["mean"]

    if group.iloc[0]["dataset"] in REGRESSION:

        group["improvement"] = base - group["mean"]

    else:

        group["improvement"] = group["mean"] - base

    return group


df = df.groupby(
["dataset","model"],
group_keys=False
).apply(compute_improvement)


########################################
# TABLE BUILDER
########################################

def build_raw_table(data,model):

    m = data[data.model==model]

    pivot = m.pivot_table(

        index="dataset",

        columns="method",

        values=["mean","std","outlier_pct"],

        aggfunc="first"

    )

    table = pd.DataFrame(index=pivot.index)

    for method in m["method"].unique():

        try:

            mean = pivot["mean"][method]

            std = pivot["std"][method]

            out = pivot["outlier_pct"][method]

            table[method] = (

                mean.round(4).astype(str)
                + " ± "
                + std.round(4).astype(str)
                + " ("
                + (100*out).round(2).astype(str)
                + "%)"
            )

        except:
            table[method]="NA"

    return table


########################################
# IMPROVEMENT TABLE
########################################

def build_improvement_table(data,model):

    m = data[data.model==model]

    pivot = m.pivot_table(

        index="dataset",

        columns="method",

        values=["improvement","std","outlier_pct"],

        aggfunc="first"

    )

    table = pd.DataFrame(index=pivot.index)

    for method in m["method"].unique():

        try:

            imp = pivot["improvement"][method]

            std = pivot["std"][method]

            out = pivot["outlier_pct"][method]

            table[method] = (

                imp.round(4).astype(str)
                + " ± "
                + std.round(4).astype(str)
                + " ("
                + (100*out).round(2).astype(str)
                + "%)"
            )

        except:

            table[method]="NA"

    avg = m.groupby("method")["improvement"].mean()

    table.loc["AVERAGE"] = avg

    return table


########################################
# METHOD SUMMARY
########################################

def method_summary(data):

    rows=[]

    for method in data.method.unique():

        sub = data[data.method==method]

        improved = (sub.improvement>0).sum()

        total = len(sub)

        avg_imp = sub.improvement.mean()

        std_imp = sub.improvement.std()

        avg_out = sub.outlier_pct.mean()

        corr = sub[["improvement","outlier_pct"]].corr().iloc[0,1]

        rows.append({

            "method":method,

            "improved":

            f"{improved}/{total}",

            "avg_improvement":avg_imp,

            "std_improvement":std_imp,

            "avg_outliers":avg_out,

            "correlation":corr

        })

    return pd.DataFrame(rows)


########################################
# SCATTER
########################################

def scatter(data,model,folder):

    m=data[data.model==model]

    plt.figure(figsize=(8,6))

    for method in m.method.unique():

        sub=m[m.method==method]

        plt.scatter(

            sub.outlier_pct*100,

            sub.improvement,

            label=method

        )

    plt.axhline(0,color="black")

    plt.xlabel("Outliers %")

    plt.ylabel("Performance change")

    plt.title(model)

    plt.legend(fontsize=7)

    plt.savefig(

        f"{folder}/{model}_scatter.png",

        dpi=200,

        bbox_inches="tight"

    )

    plt.close()


########################################
# MAIN EXPORT
########################################

def export(group,name):

    base=f"analysis/{name}"

    os.makedirs(base+"/tables",exist_ok=True)

    os.makedirs(base+"/plots",exist_ok=True)

    for model in MODELS:

        raw=build_raw_table(group,model)

        raw.to_csv(

            f"{base}/tables/{model}_raw.csv"
        )

        imp=build_improvement_table(group,model)

        imp.to_csv(

            f"{base}/tables/{model}_improvement.csv"
        )

        scatter(group,model,base+"/plots")

    summary=method_summary(group)

    summary.to_csv(

        f"{base}/tables/method_summary.csv",

        index=False
    )


########################################
# RUN
########################################

export(
df[df.dataset.isin(CLASSIFICATION)],
"classification"
)

export(
df[df.dataset.isin(REGRESSION)],
"regression"
)

print("Analysis complete")
```

```python
########################################
# EXTRA TABLES FOR THESIS
########################################

os.makedirs("analysis/extra_tables",exist_ok=True)

########################################
# TABLE 1 – MODEL COMPARISON PER METHOD
########################################

def build_model_comparison(df):

    pivot = df.pivot_table(

        index="dataset",

        columns=["method","model"],

        values="mean",

        aggfunc="first"

    )

    table = pd.DataFrame(index=pivot.index)

    methods = sorted(df["method"].unique())

    models = ["mlp","randomforest","xgboost"]

    for method in methods:

        for model in models:

            col_name=f"{method}_{model}"

            try:

                table[col_name]=pivot[(method,model)].round(4)

            except:

                table[col_name]="NA"

    return table.sort_index()


########################################
# TABLE 2 – BASELINES + OUTLIER %
########################################

def build_baseline_outliers(df):

    datasets=sorted(df.dataset.unique())

    table=pd.DataFrame(index=datasets)

    models=["mlp","randomforest","xgboost"]

    ################################
    # BASELINES (None_remove)
    ################################

    baseline=df[
        (df.detector.isna())|
        (df.detector=="None")
    ]

    for model in models:

        sub=baseline[baseline.model==model]

        try:

            table[f"{model}_baseline"]=(
                sub.set_index("dataset")["mean"]
                .round(4)
            )

        except:

            table[f"{model}_baseline"]="NA"

    ################################
    # OUTLIER %
    ################################

    methods=sorted(df.method.unique())

    for method in methods:

        # outlier % identical across models
        sub=df[
            (df.method==method)&
            (df.model=="mlp")
        ]

        try:

            table[f"{method}_outlier_pct"]=(
                sub.set_index("dataset")["outlier_pct"]
                *100
            ).round(2)

        except:

            table[f"{method}_outlier_pct"]="NA"

    return table.sort_index()


########################################
# BUILD TABLES
########################################

table1=build_model_comparison(df)

table2=build_baseline_outliers(df)

########################################
# EXPORT
########################################

table1.to_csv(
"analysis/extra_tables/table1_model_comparison.csv"
)

table2.to_csv(
"analysis/extra_tables/table2_baseline_outliers.csv"
)

print("Extra tables created")
```

```python
########################################
# EXTRA TABLES FOR THESIS (UPDATED)
########################################

os.makedirs("analysis/extra_tables",exist_ok=True)

########################################
# TABLE 1 – MODEL COMPARISON (READABLE)
########################################

def build_model_comparison(df):

    pivot = df.pivot_table(

        index="dataset",

        columns=["method","model"],

        values="mean",

        aggfunc="first"

    )

    table = pd.DataFrame(index=pivot.index)

    methods = sorted(df["method"].unique())

    models = ["mlp","randomforest","xgboost"]

    for method in methods:

        for model in models:

            col_name=f"{method}_{model}"

            try:

                table[col_name]=pivot[(method,model)].round(4)

            except:

                table[col_name]="NA"

        # spacer column for readability
        table[f"{method}_sep"]=""

    return table.sort_index()


########################################
# TABLE 2 – BASELINES + PERCENTILE ONLY
########################################

def build_baseline_outliers(df):

    datasets=sorted(df.dataset.unique())

    table=pd.DataFrame(index=datasets)

    models=["mlp","randomforest","xgboost"]

    ################################
    # BASELINES
    ################################

    baseline=df[
        (df.detector.isna())|
        (df.detector=="None")
    ]

    for model in models:

        try:

            sub=baseline[baseline.model==model]

            table[f"{model}_baseline"]=(
                sub.set_index("dataset")["mean"]
                .round(4)
            )

        except:

            table[f"{model}_baseline"]="NA"

    ################################
    # PERCENTILE OUTLIERS ONLY
    ################################

    percentile_methods=[
        "Percentile_remove",
        "Percentile_clip"
    ]

    for method in percentile_methods:

        sub=df[
            (df.method==method)&
            (df.model=="mlp")
        ]

        try:

            table[f"{method}_outlier_pct"]=(
                sub.set_index("dataset")["outlier_pct"]
                *100
            ).round(2)

        except:

            table[f"{method}_outlier_pct"]="NA"

    return table.sort_index()


########################################
# BUILD
########################################

table1=build_model_comparison(df)

table2=build_baseline_outliers(df)

########################################
# EXPORT
########################################

table1.to_csv(
"analysis/extra_tables/table1_model_comparison.csv"
)

table2.to_csv(
"analysis/extra_tables/table2_percentile_outliers.csv"
)

print("Extra tables created")
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import spearmanr

INPUT="analysis/extra_tables/table2_diminished.csv"

os.makedirs("analysis/extra_tables",exist_ok=True)
os.makedirs("analysis/extra_plots",exist_ok=True)

########################################
# LOAD
########################################

df=pd.read_csv(INPUT,index_col=0)

########################################
# COMPUTE GAPS
########################################

df["rf_gap"]=df["randomforest_baseline"]-df["mlp_baseline"]

df["xgb_gap"]=df["xgboost_baseline"]-df["mlp_baseline"]

########################################
# CORRELATIONS
########################################

rf_corr,_=spearmanr(
    df["rf_gap"],
    df["Percentile_remove_outlier_pct"]
)

xgb_corr,_=spearmanr(
    df["xgb_gap"],
    df["Percentile_remove_outlier_pct"]
)

########################################
# ADD ROW TO TABLE
########################################

corr_row={

    "mlp_baseline":"",

    "randomforest_baseline":round(rf_corr,4),

    "xgboost_baseline":round(xgb_corr,4),

    "Percentile_remove_outlier_pct":""

}

df.loc["SPEARMAN_CORRELATION"]=corr_row

########################################
# SAVE TABLE
########################################

df.to_csv(
"analysis/extra_tables/table2_with_correlation.csv"
)

########################################
# SCATTER RF
########################################

plt.figure(figsize=(7,5))

plt.scatter(
    df.iloc[:-1]["Percentile_remove_outlier_pct"],
    df.iloc[:-1]["rf_gap"]
)

for dataset in df.index[:-1]:

    plt.text(
        df.loc[dataset,"Percentile_remove_outlier_pct"],
        df.loc[dataset,"rf_gap"],
        dataset,
        fontsize=7
    )

plt.xlabel("Outlier %")
plt.ylabel("RF - MLP performance gap")

plt.title(
f"RF vs MLP gap vs outliers (rho={rf_corr:.3f})"
)

plt.axhline(0,color="black")

plt.savefig(
"analysis/extra_plots/rf_gap_scatter.png",
dpi=200,
bbox_inches="tight"
)

plt.close()

########################################
# SCATTER XGB
########################################

plt.figure(figsize=(7,5))

plt.scatter(
    df.iloc[:-1]["Percentile_remove_outlier_pct"],
    df.iloc[:-1]["xgb_gap"]
)

for dataset in df.index[:-1]:

    plt.text(
        df.loc[dataset,"Percentile_remove_outlier_pct"],
        df.loc[dataset,"xgb_gap"],
        dataset,
        fontsize=7
    )

plt.xlabel("Outlier %")
plt.ylabel("XGB - MLP performance gap")

plt.title(
f"XGB vs MLP gap vs outliers (rho={xgb_corr:.3f})"
)

plt.axhline(0,color="black")

plt.savefig(
"analysis/extra_plots/xgb_gap_scatter.png",
dpi=200,
bbox_inches="tight"
)

plt.close()

print("Correlation table and plots created")
```
