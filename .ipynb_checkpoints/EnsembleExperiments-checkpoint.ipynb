```python
import os
import json
import itertools
import warnings
import numpy as np
import pandas as pd

from pathlib import Path

from sklearn.model_selection import (
    KFold,
    RepeatedKFold
)

from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    mean_squared_error
)

from xgboost import (
    XGBClassifier,
    XGBRegressor
)

import torch
import torch.nn as nn
from torch.utils.data import (
    Dataset,
    DataLoader,
    TensorDataset
)

from tqdm.auto import tqdm

warnings.filterwarnings("ignore")

DEVICE = (
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

print("DEVICE =", DEVICE)
```

```python
DATA_PATH = "data"

OUTPUT_PATH = "results_hpo"

PRED_PATH = f"{OUTPUT_PATH}/predictions"

os.makedirs(
    OUTPUT_PATH,
    exist_ok=True
)

os.makedirs(
    PRED_PATH,
    exist_ok=True
)

DATASETS = [

    "Bank_Customer_Churn_Dataset",
    "credit-g",
    "phoneme",

    "california_housing",
    "house_16H",
    "wine_quality"

]

N_SPLITS = 5
N_REPEATS = 2

TOP_K = 3

SEED = 42
```

```python
def load_json(path):

    return json.loads(
        Path(path).read_text()
    )


def load_dataset(dataset):

    base = Path(DATA_PATH)/dataset

    def block(name):

        return {

            x: np.load(
                base/f"{name}_{x}.npy",
                allow_pickle=True
            )

            for x in ["train","test"]

        }

    N = (
        block("N")
        if (base/"N_train.npy").exists()
        else None
    )

    C = (
        block("C")
        if (base/"C_train.npy").exists()
        else None
    )

    y = block("y")

    info = load_json(
        base/"info.json"
    )

    return N,C,y,info
```

```python
def encode_train_test(
    X_train,
    X_test
):

    X_train = pd.DataFrame(X_train)
    X_test  = pd.DataFrame(X_test)

    encoders = {}

    for col in X_train.columns:

        if X_train[col].dtype == object:

            le = LabelEncoder()

            X_train[col] = le.fit_transform(
                X_train[col].astype(str)
            )

            X_test[col] = X_test[col].astype(str)

            X_test[col] = X_test[col].map(

                lambda x:

                le.transform([x])[0]

                if x in le.classes_

                else 0

            )

            encoders[col] = le

    return (
        X_train.values,
        X_test.values
    )


def preprocess_train(X):

    X = pd.DataFrame(X)

    means = {}
    stds = {}

    for col in X.columns:

        if np.issubdtype(
            X[col].dtype,
            np.number
        ):

            m = X[col].mean()

            s = X[col].std()

            if s == 0:
                s = 1

            X[col] = (
                X[col]
                .fillna(m)
            )

            X[col] = (
                X[col]-m
            )/s

            means[col] = m
            stds[col] = s

    X = X.fillna(0)

    return (
        X.values.astype(np.float32),
        means,
        stds
    )


def preprocess_test(
    X,
    means,
    stds
):

    X = pd.DataFrame(X)

    for col in means:

        if col in X:

            X[col] = (
                X[col]
                .fillna(means[col])
            )

            X[col] = (
                X[col]-means[col]
            )/stds[col]

    X = X.fillna(0)

    return X.values.astype(np.float32)
```

```python
class RealMLP(nn.Module):

    def __init__(
        self,
        input_dim,
        output_dim,
        task,
        n_layers=3,
        width=256,
        dropout=0.0
    ):

        super().__init__()

        layers = []

        d = input_dim

        for _ in range(n_layers):

            layers.append(
                nn.Linear(d, width)
            )

            layers.append(
                nn.Mish()
            )

            if dropout > 0:

                layers.append(
                    nn.Dropout(dropout)
                )

            d = width

        layers.append(
            nn.Linear(d, output_dim)
        )

        self.net = nn.Sequential(*layers)

        self.task = task

    def forward(self, x):

        return self.net(x)
```

```python
def train_realmlp(

    X_train,
    y_train,

    X_val,
    y_val,

    task,

    config,

    max_epochs=50,
    patience=10

):

    n_layers = config["n_layers"]
    width = config["width"]
    dropout = config["dropout"]

    X_train_t = torch.tensor(
        X_train,
        dtype=torch.float32
    )

    X_val_t = torch.tensor(
        X_val,
        dtype=torch.float32
    )

    if task == "classification":

        y_train_t = torch.tensor(
            y_train,
            dtype=torch.long
        )

        y_val_t = torch.tensor(
            y_val,
            dtype=torch.long
        )

        output_dim = len(
            np.unique(y_train)
        )

        loss_fn = nn.CrossEntropyLoss()

    else:

        y_train_t = torch.tensor(
            y_train,
            dtype=torch.float32
        ).view(-1,1)

        y_val_t = torch.tensor(
            y_val,
            dtype=torch.float32
        ).view(-1,1)

        output_dim = 1

        loss_fn = nn.MSELoss()

    model = RealMLP(

        input_dim=X_train.shape[1],
        output_dim=output_dim,

        task=task,

        n_layers=n_layers,
        width=width,
        dropout=dropout

    ).to(DEVICE)

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=1e-3,
        weight_decay=1e-4
    )

    train_loader = DataLoader(

        TensorDataset(
            X_train_t,
            y_train_t
        ),

        batch_size=256,
        shuffle=True

    )

    best_loss = np.inf
    best_state = None

    wait = 0

    for epoch in range(max_epochs):

        model.train()

        for xb,yb in train_loader:

            xb = xb.to(DEVICE)
            yb = yb.to(DEVICE)

            pred = model(xb)

            loss = loss_fn(
                pred,
                yb
            )

            optimizer.zero_grad()

            loss.backward()

            optimizer.step()

        model.eval()

        with torch.no_grad():

            pred_val = model(
                X_val_t.to(DEVICE)
            )

            val_loss = loss_fn(
                pred_val,
                y_val_t.to(DEVICE)
            ).item()

        if val_loss < best_loss:

            best_loss = val_loss

            best_state = {

                k:v.cpu()

                for k,v in
                model.state_dict().items()

            }

            wait = 0

        else:

            wait += 1

            if wait >= patience:
                break

    model.load_state_dict(
        best_state
    )

    return model
```

```python
def predict_realmlp(
    model,
    X,
    task
):

    X_t = torch.tensor(
        X,
        dtype=torch.float32
    ).to(DEVICE)

    model.eval()

    with torch.no_grad():

        pred = model(X_t)

        if task == "classification":

            prob = torch.softmax(
                pred,
                dim=1
            )

            return (
                prob.cpu().numpy()
            )

        return (
            pred.cpu()
            .numpy()
            .ravel()
        )
```

```python
REALMLP_DEFAULT = {

    "n_layers":3,
    "width":256,
    "dropout":0.0

}
```

```python
REALMLP_CONFIGS = []

for n_layers in [2,3,4]:

    for width in [256,384,512]:

        for dropout in [0.0,0.2]:

            REALMLP_CONFIGS.append({

                "n_layers":n_layers,
                "width":width,
                "dropout":dropout

            })

print(
    "RealMLP configs:",
    len(REALMLP_CONFIGS)
)
```

```python
XGB_DEFAULT = {

    "max_depth": 6,
    "learning_rate": 0.1,

    "n_estimators": 200,

    "subsample": 1.0,
    "colsample_bytree": 1.0

}
```

```python
XGB_CONFIGS = []

for depth in [4,6,8]:

    for lr in [0.05,0.1]:

        for subsample in [0.8,1.0]:

            for colsample in [0.8,1.0]:

                XGB_CONFIGS.append({

                    "max_depth": depth,

                    "learning_rate": lr,

                    "n_estimators": 200,

                    "subsample": subsample,

                    "colsample_bytree": colsample

                })

print(
    "XGB configs:",
    len(XGB_CONFIGS)
)
```

```python
def build_xgb(
    config,
    task
):

    if task == "classification":

        return XGBClassifier(

            eval_metric="logloss",

            tree_method="hist",

            random_state=SEED,

            n_jobs=-1,

            verbosity=0,

            **config

        )

    return XGBRegressor(

        tree_method="hist",

        random_state=SEED,

        n_jobs=-1,

        verbosity=0,

        **config

    )
```

```python
def evaluate_config(

    X,
    y,

    task,

    model_name,
    config

):

    cv = KFold(

        n_splits=3,

        shuffle=True,

        random_state=SEED

    )

    scores = []

    for tr,val in cv.split(X):

        X_tr = X[tr]
        X_val = X[val]

        y_tr = y[tr]
        y_val = y[val]

        if model_name == "realmlp":

            model = train_realmlp(

                X_tr,
                y_tr,

                X_val,
                y_val,

                task,
                config

            )

            pred = predict_realmlp(

                model,
                X_val,
                task

            )

            if task == "classification":

                pred = pred.argmax(1)

                score = accuracy_score(
                    y_val,
                    pred
                )

            else:

                score = -mean_squared_error(
                    y_val,
                    pred
                )

        else:

            model = build_xgb(
                config,
                task
            )

            model.fit(
                X_tr,
                y_tr
            )

            pred = model.predict(
                X_val
            )

            if task == "classification":

                score = accuracy_score(
                    y_val,
                    pred
                )

            else:

                score = -mean_squared_error(
                    y_val,
                    pred
                )

        scores.append(score)

    return np.mean(scores)
```

```python
def find_best_configs(

    X,
    y,

    task,

    model_name,

    top_k=3

):

    if model_name == "realmlp":

        configs = REALMLP_CONFIGS

    else:

        configs = XGB_CONFIGS

    rows = []

    for cfg in tqdm(
        configs,
        leave=False
    ):

        score = evaluate_config(

            X,
            y,

            task,

            model_name,

            cfg

        )

        rows.append({

            "config": cfg,
            "score": score

        })

    df = pd.DataFrame(rows)

    df = df.sort_values(

        "score",

        ascending=False

    )

    return list(
        df.head(top_k)["config"]
    ), df
```

```python
def fit_ensemble(

    X_train,
    y_train,

    task,

    model_name,

    configs

):

    models = []

    for cfg in configs:

        if model_name == "realmlp":

            model = train_realmlp(

                X_train,
                y_train,

                X_train,
                y_train,

                task,
                cfg

            )

        else:

            model = build_xgb(
                cfg,
                task
            )

            model.fit(
                X_train,
                y_train
            )

        models.append(model)

    return models
```

```python
def ensemble_predict(

    models,

    X,

    task,

    model_name

):

    preds = []

    for model in models:

        if model_name == "realmlp":

            p = predict_realmlp(
                model,
                X,
                task
            )

        else:

            if task == "classification":

                p = model.predict_proba(X)

            else:

                p = model.predict(X)

        preds.append(p)

    if task == "classification":

        mean_prob = np.mean(
            preds,
            axis=0
        )

        return mean_prob.argmax(1)

    return np.mean(
        preds,
        axis=0
    )
```

```python
def build_X_y(dataset):

    N,C,y,info = load_dataset(dataset)

    if N is not None and C is not None:

        X = np.hstack([
            N["train"],
            C["train"]
        ])

    elif N is not None:

        X = N["train"]

    else:

        X = C["train"]

    y = y["train"].ravel()

    X = pd.DataFrame(X)

    for col in X.columns:

        if X[col].dtype == object:

            le = LabelEncoder()

            X[col] = le.fit_transform(
                X[col].astype(str)
            )

    X = X.values

    task = (
        "classification"
        if info["task_type"] != "regression"
        else "regression"
    )

    if task == "classification":

        le = LabelEncoder()

        y = le.fit_transform(y)

    return X,y,task
```

```python
def evaluate_default_model(

    X_train,
    y_train,

    X_test,
    y_test,

    task,
    model_name

):

    if model_name == "realmlp":

        model = train_realmlp(

            X_train,
            y_train,

            X_train,
            y_train,

            task,

            REALMLP_DEFAULT

        )

        pred = predict_realmlp(
            model,
            X_test,
            task
        )

        if task == "classification":
            pred = pred.argmax(1)

    else:

        model = build_xgb(
            XGB_DEFAULT,
            task
        )

        model.fit(
            X_train,
            y_train
        )

        pred = model.predict(
            X_test
        )

    if task == "classification":

        return accuracy_score(
            y_test,
            pred
        )

    return mean_squared_error(
        y_test,
        pred
    )
```

```python
def evaluate_dataset(dataset):

    print()
    print("="*60)
    print(dataset)
    print("="*60)

    X,y,task = build_X_y(
        dataset
    )

    print("Task:",task)

    model_cache = {}

    rows = []

    fold_id = 0
    
    for run_id in range(N_REPEATS):
    
        kf = KFold(
    
            n_splits=N_SPLITS,
    
            shuffle=True,
    
            random_state=SEED + run_id
    
        )
    
        for fold_idx,(train_idx,test_idx) in enumerate(kf.split(X)):
    
            fold_id += 1

        fold_id += 1

        X_train = X[train_idx]
        X_test  = X[test_idx]

        y_train = y[train_idx]
        y_test  = y[test_idx]

        X_train,means,stds = preprocess_train(
            X_train
        )

        X_test = preprocess_test(
            X_test,
            means,
            stds
        )

        ##################################################
        # HPO ONCE PER DATASET
        ##################################################

        if "realmlp" not in model_cache:

            print("Searching RealMLP configs...")

            best_cfgs,df_cfg = find_best_configs(

                X_train,
                y_train,

                task,

                "realmlp",

                TOP_K

            )

            model_cache["realmlp"] = best_cfgs

            df_cfg.to_csv(

                f"{OUTPUT_PATH}/{dataset}_realmlp_configs.csv",

                index=False

            )

        if "xgb" not in model_cache:

            print("Searching XGB configs...")

            best_cfgs,df_cfg = find_best_configs(

                X_train,
                y_train,

                task,

                "xgb",

                TOP_K

            )

            model_cache["xgb"] = best_cfgs

            df_cfg.to_csv(

                f"{OUTPUT_PATH}/{dataset}_xgb_configs.csv",

                index=False

            )

        ##################################################
        # DEFAULTS
        ##################################################

        score = evaluate_default_model(

            X_train,
            y_train,

            X_test,
            y_test,

            task,

            "realmlp"

        )

        rows.append({

            "dataset":dataset,

            "model":"realmlp_default",

            "fold":fold_id,

            "score":score

        })

        score = evaluate_default_model(

            X_train,
            y_train,

            X_test,
            y_test,

            task,

            "xgb"

        )

        rows.append({

            "dataset":dataset,

            "model":"xgb_default",

            "fold":fold_id,

            "score":score

        })

        ##################################################
        # ENSEMBLES
        ##################################################

        ensemble = fit_ensemble(

            X_train,
            y_train,

            task,

            "realmlp",

            model_cache["realmlp"]

        )

        pred = ensemble_predict(

            ensemble,

            X_test,

            task,

            "realmlp"

        )

        if task == "classification":

            score = accuracy_score(
                y_test,
                pred
            )

        else:

            score = mean_squared_error(
                y_test,
                pred
            )

        rows.append({

            "dataset":dataset,

            "model":"realmlp_ensemble",

            "fold":fold_id,

            "score":score

        })

        ensemble = fit_ensemble(

            X_train,
            y_train,

            task,

            "xgb",

            model_cache["xgb"]

        )

        pred = ensemble_predict(

            ensemble,

            X_test,

            task,

            "xgb"

        )

        if task == "classification":

            score = accuracy_score(
                y_test,
                pred
            )

        else:

            score = mean_squared_error(
                y_test,
                pred
            )

        rows.append({

            "dataset":dataset,

            "model":"xgb_ensemble",

            "fold":fold_id,

            "score":score

        })

    return pd.DataFrame(rows)
```

```python
def evaluate_dataset(dataset):

    print()
    print("="*60)
    print(dataset)
    print("="*60)

    X,y,task = build_X_y(
        dataset
    )

    print("Task:",task)

    model_cache = {}

    rows = []

    rkf = RepeatedKFold(

        n_splits=N_SPLITS,

        n_repeats=N_REPEATS,

        random_state=SEED

    )

    fold_id = 0

    for train_idx,test_idx in tqdm(

        rkf.split(X),

        total=N_SPLITS*N_REPEATS

    ):

        fold_id += 1

        X_train = X[train_idx]
        X_test  = X[test_idx]

        y_train = y[train_idx]
        y_test  = y[test_idx]

        X_train,means,stds = preprocess_train(
            X_train
        )

        X_test = preprocess_test(
            X_test,
            means,
            stds
        )

        ##################################################
        # HPO ONCE PER DATASET
        ##################################################

        if "realmlp" not in model_cache:

            print("Searching RealMLP configs...")

            best_cfgs,df_cfg = find_best_configs(

                X_train,
                y_train,

                task,

                "realmlp",

                TOP_K

            )

            model_cache["realmlp"] = best_cfgs

            df_cfg.to_csv(

                f"{OUTPUT_PATH}/{dataset}_realmlp_configs.csv",

                index=False

            )

        if "xgb" not in model_cache:

            print("Searching XGB configs...")

            best_cfgs,df_cfg = find_best_configs(

                X_train,
                y_train,

                task,

                "xgb",

                TOP_K

            )

            model_cache["xgb"] = best_cfgs

            df_cfg.to_csv(

                f"{OUTPUT_PATH}/{dataset}_xgb_configs.csv",

                index=False

            )

        ##################################################
        # DEFAULTS
        ##################################################

        score = evaluate_default_model(

            X_train,
            y_train,

            X_test,
            y_test,

            task,

            "realmlp"

        )

        rows.append({

            "dataset":dataset,

            "model":"realmlp_default",

            "fold":fold_id,

            "score":score

        })

        score = evaluate_default_model(

            X_train,
            y_train,

            X_test,
            y_test,

            task,

            "xgb"

        )

        rows.append({

            "dataset":dataset,

            "model":"xgb_default",

            "fold":fold_id,

            "score":score

        })

        ##################################################
        # ENSEMBLES
        ##################################################

        ensemble = fit_ensemble(

            X_train,
            y_train,

            task,

            "realmlp",

            model_cache["realmlp"]

        )

        pred = ensemble_predict(

            ensemble,

            X_test,

            task,

            "realmlp"

        )

        if task == "classification":

            score = accuracy_score(
                y_test,
                pred
            )

        else:

            score = mean_squared_error(
                y_test,
                pred
            )

        rows.append({

            "dataset":dataset,

            "model":"realmlp_ensemble",

            "fold":fold_id,

            "score":score

        })

        ensemble = fit_ensemble(

            X_train,
            y_train,

            task,

            "xgb",

            model_cache["xgb"]

        )

        pred = ensemble_predict(

            ensemble,

            X_test,

            task,

            "xgb"

        )

        if task == "classification":

            score = accuracy_score(
                y_test,
                pred
            )

        else:

            score = mean_squared_error(
                y_test,
                pred
            )

        rows.append({

            "dataset":dataset,

            "model":"xgb_ensemble",

            "fold":fold_id,

            "score":score

        })

    return pd.DataFrame(rows)
```

```python
def run_all():

    all_results = []

    for dataset in DATASETS:

        df = evaluate_dataset(
            dataset
        )

        all_results.append(df)

        current = pd.concat(
            all_results,
            ignore_index=True
        )

        current.to_csv(

            f"{OUTPUT_PATH}/all_results.csv",

            index=False

        )

    final_df = pd.concat(

        all_results,

        ignore_index=True

    )

    return final_df
```

```python
def summarize_results(df):

    summary = (

        df.groupby(

            ["dataset","model"]

        )

        ["score"]

        .agg(

            ["mean","std"]

        )

        .reset_index()

    )

    summary.to_csv(

        f"{OUTPUT_PATH}/summary.csv",

        index=False

    )

    return summary
```

```python
df = run_all()

summary = summarize_results(df)

summary
```
