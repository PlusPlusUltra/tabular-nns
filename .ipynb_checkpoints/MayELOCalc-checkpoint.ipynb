```python
# elo_analysis.py

import os
import numpy as np
import pandas as pd
from scipy.stats import wilcoxon
from itertools import combinations

########################################
# CONFIG
########################################

RESULT_PATH = "results_cv/all_results.csv"

OUTPUT_DIR = "tables/elo"
os.makedirs(OUTPUT_DIR, exist_ok=True)

K = 32  # Elo update strength


########################################
# HELPERS
########################################

def expected_score(r1, r2):
    return 1 / (1 + 10 ** ((r2 - r1) / 400))


def update_elo(r1, r2, s1, s2):
    e1 = expected_score(r1, r2)
    e2 = expected_score(r2, r1)

    r1_new = r1 + K * (s1 - e1)
    r2_new = r2 + K * (s2 - e2)

    return r1_new, r2_new


def wilcoxon_result(x, y, higher_is_better=True):

    try:
        stat, p = wilcoxon(x, y)
    except:
        return 0.5  # draw

    if p >= 0.05:
        return 0.5

    if higher_is_better:
        return 1 if np.mean(x) > np.mean(y) else 0
    else:
        return 1 if np.mean(x) < np.mean(y) else 0


########################################
# LOAD DATA
########################################

df = pd.read_csv(RESULT_PATH)

# define task type
REG_DATASETS = [
    "california_housing", "cpu_act",
    "Bike_Sharing_Demand", "abalone", "wine_quality"
]

df["task"] = df["dataset"].apply(
    lambda x: "regression" if x in REG_DATASETS else "classification"
)


########################################
# CORE FUNCTION
########################################

def compute_elo(methods_df, base_rating=None):

    methods = methods_df["method"].unique()
    elo = {m: 1000 for m in methods}

    if base_rating:
        elo.update(base_rating)

    datasets = methods_df["dataset"].unique()

    for dataset in datasets:

        dset = methods_df[methods_df["dataset"] == dataset]
        task = dset["task"].iloc[0]

        higher_is_better = (task == "classification")

        for m1, m2 in combinations(methods, 2):

            x = dset[dset["method"] == m1]["score"].values
            y = dset[dset["method"] == m2]["score"].values

            if len(x) == 0 or len(y) == 0:
                continue

            res = wilcoxon_result(x, y, higher_is_better)

            if res == 1:
                s1, s2 = 1, 0
            elif res == 0:
                s1, s2 = 0, 1
            else:
                s1, s2 = 0.5, 0.5

            elo[m1], elo[m2] = update_elo(
                elo[m1], elo[m2], s1, s2
            )

    return pd.DataFrame({
        "method": list(elo.keys()),
        "elo": list(elo.values())
    }).sort_values("elo", ascending=False)


########################################
# 1️⃣ MODELS ONLY (NO OUTLIERS)
########################################

df_models = df[df["detector"] == "None"].copy()

df_models["method"] = df_models["model"]

elo_models = compute_elo(df_models)

elo_models.to_csv(f"{OUTPUT_DIR}/elo_models.csv", index=False)


########################################
# 2️⃣ MLP VS OUTLIERS
########################################

df_mlp = df[df["model"] == "mlp"].copy()

df_mlp["method"] = (
    df_mlp["model"] + "_" +
    df_mlp["detector"].astype(str) + "_" +
    df_mlp["treatment"]
)

# base MLP = 1000
base = {
    "mlp_None_remove": 1000
}

elo_mlp = compute_elo(df_mlp, base_rating=base)

elo_mlp.to_csv(f"{OUTPUT_DIR}/elo_mlp_outliers.csv", index=False)


########################################
# 3️⃣ ALL METHODS
########################################

df_all = df.copy()

df_all["method"] = (
    df_all["model"] + "_" +
    df_all["detector"].astype(str) + "_" +
    df_all["treatment"]
)

# RF base = 1000
base = {
    "randomforest_None_remove": 1000
}

elo_all = compute_elo(df_all, base_rating=base)

elo_all.to_csv(f"{OUTPUT_DIR}/elo_all.csv", index=False)


print("ELO tables saved in:", OUTPUT_DIR)
```

```python
import os
import numpy as np
import pandas as pd
from scipy.stats import wilcoxon
from itertools import combinations
import warnings

########################################
# CONFIG
########################################

RESULT_PATH = "results_cv/all_results.csv"
OUTPUT_DIR = "tables/elo"
os.makedirs(OUTPUT_DIR, exist_ok=True)

N_BOOTSTRAP = 200
K = 32

warnings.filterwarnings("ignore")

########################################
# HELPERS
########################################

def expected_score(r1, r2):
    return 1 / (1 + 10 ** ((r2 - r1) / 400))


def update_elo(r1, r2, s1, s2):
    e1 = expected_score(r1, r2)
    e2 = expected_score(r2, r1)
    return (
        r1 + K * (s1 - e1),
        r2 + K * (s2 - e2)
    )


def wilcoxon_result(x, y, higher_is_better=True):

    # caso identico → pareggio
    if np.allclose(x, y):
        return 0.5

    try:
        stat, p = wilcoxon(x, y)
    except:
        return 0.5

    if np.isnan(p) or p >= 0.05:
        return 0.5

    if higher_is_better:
        return 1 if np.mean(x) > np.mean(y) else 0
    else:
        return 1 if np.mean(x) < np.mean(y) else 0


########################################
# LOAD DATA
########################################

df = pd.read_csv(RESULT_PATH)

REG_DATASETS = [
    "california_housing", "cpu_act",
    "Bike_Sharing_Demand", "abalone", "wine_quality"
]

df["task"] = df["dataset"].apply(
    lambda x: "regression" if x in REG_DATASETS else "classification"
)

########################################
# CORE ELO (single run)
########################################

def compute_elo_once(methods_df, datasets_sample, base_rating=None):

    methods = methods_df["method"].unique()
    elo = {m: 1000 for m in methods}

    if base_rating:
        elo.update(base_rating)

    for dataset in datasets_sample:

        dset = methods_df[methods_df["dataset"] == dataset]
        if len(dset) == 0:
            continue

        task = dset["task"].iloc[0]
        higher_is_better = (task == "classification")

        for m1, m2 in combinations(methods, 2):

            x = dset[dset["method"] == m1]["score"].values
            y = dset[dset["method"] == m2]["score"].values

            if len(x) == 0 or len(y) == 0:
                continue

            res = wilcoxon_result(x, y, higher_is_better)

            if res == 1:
                s1, s2 = 1, 0
            elif res == 0:
                s1, s2 = 0, 1
            else:
                s1, s2 = 0.5, 0.5

            elo[m1], elo[m2] = update_elo(
                elo[m1], elo[m2], s1, s2
            )

    return elo


########################################
# BOOTSTRAP WRAPPER
########################################

def compute_elo_bootstrap(methods_df, base_rating=None):

    datasets = methods_df["dataset"].unique()
    methods = methods_df["method"].unique()

    all_elo = {m: [] for m in methods}

    for i in range(N_BOOTSTRAP):

        sampled = np.random.choice(
            datasets,
            size=len(datasets),
            replace=True
        )

        elo = compute_elo_once(
            methods_df,
            sampled,
            base_rating
        )

        for m in methods:
            all_elo[m].append(elo[m])

    rows = []

    for m in methods:
        vals = np.array(all_elo[m])

        rows.append({
            "method": m,
            "median": np.median(vals),
            "p2.5": np.percentile(vals, 2.5),
            "p97.5": np.percentile(vals, 97.5)
        })

    df_out = pd.DataFrame(rows)

    if len(df_out) == 0:
        print("WARNING: empty dataframe!")
        return df_out

    return df_out.sort_values("median", ascending=False)


########################################
# 1️⃣ MODELS ONLY (NO OUTLIERS)
########################################

df_models = df[df["detector"].isna()].copy()
df_models["method"] = df_models["model"]

elo_models = compute_elo_bootstrap(df_models)
elo_models.to_csv(f"{OUTPUT_DIR}/elo_models_bootstrap.csv", index=False)


########################################
# 2️⃣ MLP VS OUTLIERS
########################################

df_mlp = df[df["model"] == "mlp"].copy()

df_mlp["method"] = (
    df_mlp["model"] + "_" +
    df_mlp["detector"].fillna("None") + "_" +
    df_mlp["treatment"]
)

base = {"mlp_None_remove": 1000}

elo_mlp = compute_elo_bootstrap(df_mlp, base)
elo_mlp.to_csv(f"{OUTPUT_DIR}/elo_mlp_outliers_bootstrap.csv", index=False)


########################################
# 3️⃣ ALL METHODS
########################################

df_all = df.copy()

df_all["method"] = (
    df_all["model"] + "_" +
    df_all["detector"].fillna("None") + "_" +
    df_all["treatment"]
)

base = {"randomforest_None_remove": 1000}

elo_all = compute_elo_bootstrap(df_all, base)
elo_all.to_csv(f"{OUTPUT_DIR}/elo_all_bootstrap.csv", index=False)


print("✅ Bootstrap ELO tables saved in:", OUTPUT_DIR)
```

```python
# performance_tables.py

import os
import pandas as pd
import numpy as np

########################################
# CONFIG
########################################

RESULT_PATH = "results_cv/all_results.csv"
OUTPUT_DIR = "tables/performance"
os.makedirs(OUTPUT_DIR, exist_ok=True)

########################################
# LOAD
########################################

df = pd.read_csv(RESULT_PATH)

REG_DATASETS = [
    "california_housing", "cpu_act",
    "Bike_Sharing_Demand", "abalone", "wine_quality"
]

df["task"] = df["dataset"].apply(
    lambda x: "regression" if x in REG_DATASETS else "classification"
)

########################################
# HELPER
########################################

def aggregate(df_subset):
    return (
        df_subset
        .groupby(["dataset", "method"])["score"]
        .agg(["mean", "std"])
        .reset_index()
    )

########################################
# 1️⃣ BASE MODELS
########################################

def base_models_table(task):

    df_task = df[df["task"] == task].copy()

    # solo no outliers
    df_base = df_task[df_task["detector"].isna()].copy()

    df_base["method"] = df_base["model"]

    agg = aggregate(df_base)

    # pivot → dataset x model
    table = agg.pivot(
        index="dataset",
        columns="method",
        values="mean"
    )

    std_table = agg.pivot(
        index="dataset",
        columns="method",
        values="std"
    )

    # format mean ± std
    final = table.copy()

    for col in table.columns:
        final[col] = (
            table[col].round(4).astype(str)
            + " ± "
            + std_table[col].round(4).astype(str)
        )

    final.to_csv(
        f"{OUTPUT_DIR}/base_models_{task}.csv"
    )

########################################
# 2️⃣ OUTLIERS PER MODEL
########################################

def outlier_table(task, model_name):

    df_task = df[df["task"] == task].copy()

    df_model = df_task[df_task["model"] == model_name].copy()

    df_model["detector"] = df_model["detector"].fillna("None")

    df_model["method"] = (
        df_model["detector"] + "_" + df_model["treatment"]
    )

    agg = aggregate(df_model)

    table = agg.pivot(
        index="dataset",
        columns="method",
        values="mean"
    )

    std_table = agg.pivot(
        index="dataset",
        columns="method",
        values="std"
    )

    final = table.copy()

    for col in table.columns:
        final[col] = (
            table[col].round(4).astype(str)
            + " ± "
            + std_table[col].round(4).astype(str)
        )

    final.to_csv(
        f"{OUTPUT_DIR}/{model_name}_outliers_{task}.csv"
    )

########################################
# RUN
########################################

for task in ["classification", "regression"]:

    print(f"Processing {task}")

    # base models
    base_models_table(task)

    # per modello
    for model in ["mlp", "randomforest", "xgboost"]:
        outlier_table(task, model)

print("✅ Performance tables saved in:", OUTPUT_DIR)
```
