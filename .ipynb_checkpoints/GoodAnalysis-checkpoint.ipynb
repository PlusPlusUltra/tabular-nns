```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

RESULTS="results/summary.csv"

CLASSIFICATION=[
"Bank_Customer_Churn_Dataset",
"cmc",
"connect-4",
"credit-g",
"electricity",
"eucalyptus",
"eye_movements",
"house_16H",
"kc1",
"phoneme",
"splice",
"vehicle"
]

REGRESSION=[
"elevators",
"socmob",
"california_housing",
"cpu_act",
"Bike_Sharing_Demand",
"abalone",
"wine_quality"
]

########################################
# LOAD + CLEAN
########################################

df=pd.read_csv(RESULTS)

df["detector"]=df["detector"].fillna("None")

df["detector"]=df["detector"].astype(str)

df["detector"]=df["detector"].replace(
["nan","None",""],
"None"
)

df["method"]=df["detector"]+"_"+df["treatment"]

df["mean"]=df["mean"].round(5)

# remove accidental duplicate experiment rows
df=df.drop_duplicates(
["dataset","model","detector","treatment"]
)

########################################
# HELPERS
########################################

def task_type(dataset):

    if dataset in REGRESSION:
        return "regression"

    return "classification"


def better(task,new,base):

    if task=="regression":
        return new<base

    return new>base


def gap(tree,mlp,task):

    if task=="regression":
        return mlp-tree

    return tree-mlp


########################################
# ANALYSIS FUNCTION
########################################

def analyze(group,name):

    base=f"analysis/{name}"

    os.makedirs(base+"/tables",exist_ok=True)
    os.makedirs(base+"/plots",exist_ok=True)

    methods=sorted(group.method.unique())

    models=["mlp","randomforest","xgboost"]

    datasets=sorted(group.dataset.unique())

########################################
# TABLE 1 – PER MODEL METHODS
########################################

    for model in models:

        sub=group[group.model==model]

        pivot=sub.pivot_table(

            index="dataset",
            columns="method",
            values="mean",
            aggfunc="first"

        )

        pivot=pivot.reindex(datasets)

        pivot=pivot.round(5)

        if "None_remove" not in pivot.columns:
            continue

        baseline=pivot["None_remove"]

        improve_row={}

        for method in pivot.columns:

            if method=="None_remove":

                improve_row[method]="-"
                continue

            count=0
            total=len(pivot)

            for d in pivot.index:

                if pd.isna(pivot.loc[d,method]):
                    continue

                if better(

                    task_type(d),

                    pivot.loc[d,method],

                    baseline[d]

                ):
                    count+=1

            improve_row[method]=f"{count}/{total}"

        pivot.loc["IMPROVED"]=improve_row

        pivot.to_csv(
            f"{base}/tables/{model}_methods.csv"
        )

########################################
# TABLE 2 – GAP TABLES
########################################

    for tree in ["randomforest","xgboost"]:

        table=pd.DataFrame(index=datasets)

        for method in methods:

            vals=[]

            for d in datasets:

                mlp=group[
                    (group.dataset==d)&
                    (group.model=="mlp")&
                    (group.method==method)
                ]["mean"]

                tree_val=group[
                    (group.dataset==d)&
                    (group.model==tree)&
                    (group.method==method)
                ]["mean"]

                if len(mlp)==0 or len(tree_val)==0:

                    vals.append(np.nan)

                else:

                    vals.append(

                        round(

                            gap(
                                tree_val.iloc[0],
                                mlp.iloc[0],
                                task_type(d)
                            ),

                            5
                        )

                    )

            table[method]=vals

        if "None_remove" not in table.columns:
            continue

        base_gap=table["None_remove"]

        reduce_row={}

        for method in table.columns:

            if method=="None_remove":

                reduce_row[method]="-"
                continue

            count=0

            for d in table.index:

                if pd.isna(table.loc[d,method]):
                    continue

                if abs(table.loc[d,method])<abs(base_gap[d]):

                    count+=1

            reduce_row[method]=f"{count}/{len(table)}"

        table.loc["GAP_REDUCED"]=reduce_row

        table.to_csv(

            f"{base}/tables/{tree}_gaps.csv"

        )

########################################
# TABLE 3 – BASELINES + OUTLIERS
########################################

    for detector in ["Percentile","ZScore"]:

        table=pd.DataFrame(index=datasets)

        base_rows=group[
            group.method=="None_remove"
        ]

        for model in models:

            table[model]=base_rows[
                base_rows.model==model
            ].set_index("dataset")["mean"]

        out=group[
            (group.detector==detector)&
            (group.treatment=="remove")
        ]

        out=out.drop_duplicates("dataset")

        table["outlier_pct"]=(
            out.set_index("dataset")["outlier_pct"]
            *100
        ).round(5)

        valid = table.dropna()
        
        rf_corr = np.nan
        xgb_corr = np.nan
        
        if len(valid) > 2:
        
            rf_gap = valid["randomforest"] - valid["mlp"]
        
            xgb_gap = valid["xgboost"] - valid["mlp"]
        
            rf_corr,_ = pearsonr(
                valid["outlier_pct"],
                rf_gap
            )
        
            xgb_corr,_ = pearsonr(
                valid["outlier_pct"],
                xgb_gap
            )
        
        table.loc["PEARSON_CORR"] = [
        
            "",                     # mlp column
            round(rf_corr,5),      # RF column
            round(xgb_corr,5),     # XGB column
            ""                     # outlier %
        ]

        table.to_csv(

            f"{base}/tables/{detector}_baseline_outliers.csv"

        )

########################################
# SCATTER PLOTS
########################################

    for detector in ["Percentile","ZScore"]:

        for tree in ["randomforest","xgboost"]:

            x=[]
            y=[]

            for d in datasets:

                out=group[
                    (group.dataset==d)&
                    (group.detector==detector)&
                    (group.treatment=="remove")
                ]["outlier_pct"]

                mlp=group[
                    (group.dataset==d)&
                    (group.model=="mlp")&
                    (group.method=="None_remove")
                ]["mean"]

                tree_val=group[
                    (group.dataset==d)&
                    (group.model==tree)&
                    (group.method=="None_remove")
                ]["mean"]

                if len(out)==0:
                    continue

                x.append(out.iloc[0]*100)

                y.append(
                    gap(
                        tree_val.iloc[0],
                        mlp.iloc[0],
                        task_type(d)
                    )
                )

            plt.figure()

            plt.scatter(x,y)

            plt.axhline(0,color="black")

            plt.xlabel("Outlier %")

            plt.ylabel(f"{tree}-MLP gap")

            plt.title(f"{detector} vs {tree}")

            plt.savefig(

                f"{base}/plots/{detector}_{tree}_scatter.png",

                dpi=200,

                bbox_inches="tight"

            )

            plt.close()

########################################
# MLP IMPROVEMENTS PLOT
########################################

    counts={}

    for method in methods:

        if method=="None_remove":
            continue

        count=0

        for d in datasets:

            base_val=group[
                (group.dataset==d)&
                (group.model=="mlp")&
                (group.method=="None_remove")
            ]["mean"]

            new_val=group[
                (group.dataset==d)&
                (group.model=="mlp")&
                (group.method==method)
            ]["mean"]

            if len(base_val)==0 or len(new_val)==0:
                continue

            if better(

                task_type(d),

                new_val.iloc[0],

                base_val.iloc[0]

            ):

                count+=1

        counts[method]=count

    plt.figure(figsize=(10,5))

    plt.bar(counts.keys(),counts.values())

    plt.xticks(rotation=90)

    plt.ylabel("Datasets improved")

    plt.title("MLP improvements")

    plt.savefig(

        f"{base}/plots/mlp_improvements.png",

        dpi=200,

        bbox_inches="tight"

    )

    plt.close()

########################################
# GAP REDUCTION PLOTS
########################################

    for tree in ["randomforest","xgboost"]:

        counts={}

        for method in methods:

            if method=="None_remove":
                continue

            count=0

            for d in datasets:

                base_gap=gap(

                    group[
                        (group.dataset==d)&
                        (group.model==tree)&
                        (group.method=="None_remove")
                    ]["mean"].iloc[0],

                    group[
                        (group.dataset==d)&
                        (group.model=="mlp")&
                        (group.method=="None_remove")
                    ]["mean"].iloc[0],

                    task_type(d)

                )

                new_gap=gap(

                    group[
                        (group.dataset==d)&
                        (group.model==tree)&
                        (group.method==method)
                    ]["mean"].iloc[0],

                    group[
                        (group.dataset==d)&
                        (group.model=="mlp")&
                        (group.method==method)
                    ]["mean"].iloc[0],

                    task_type(d)

                )

                if abs(new_gap)<abs(base_gap):

                    count+=1

            counts[method]=count

        plt.figure(figsize=(10,5))

        plt.bar(counts.keys(),counts.values())

        plt.xticks(rotation=90)

        plt.ylabel("Gap reductions")

        plt.title(f"{tree} gap reductions")

        plt.savefig(

            f"{base}/plots/{tree}_gap_reduction.png",

            dpi=200,

            bbox_inches="tight"

        )

        plt.close()

########################################
# RUN
########################################

analyze(
df[df.dataset.isin(CLASSIFICATION)],
"classification"
)

analyze(
df[df.dataset.isin(REGRESSION)],
"regression"
)

print("Analysis complete")
```

```python
########################################
# BASELINE TABLE (NO OUTLIER DETECTION)
########################################

def baseline_table(group,name):

    base=f"analysis/{name}/tables"

    baseline=group[
        group.method=="None_remove"
    ]

    table=baseline.pivot_table(

        index="dataset",
        columns="model",
        values="mean",
        aggfunc="first"

    )

    table=table.round(5)

    table=table.sort_index()

    table.to_csv(
        f"{base}/baseline_models.csv"
    )


########################################
# BUILD BASELINE TABLES
########################################

baseline_table(
df[df.dataset.isin(CLASSIFICATION)],
"classification"
)

baseline_table(
df[df.dataset.isin(REGRESSION)],
"regression"
)

print("Baseline tables created")
```
